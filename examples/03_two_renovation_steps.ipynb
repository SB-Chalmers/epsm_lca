# Windows first, insulation second

Compare each reached building over the same 30 years. These are comparisons, not a dated operating schedule. Do not add energy impacts from the different rows.

```python
from teaching_case import (
    make_baseline,
    renovation_options,
    show_table,
    supplied_energy,
    teaching_settings,
)

from epsm_lca import assess_renovation

baseline = make_baseline()
windows, insulation = renovation_options(baseline)
step_1 = baseline.renovate(windows)
step_2 = step_1.renovate(insulation)
settings = teaching_settings(step_2)
energy = supplied_energy(baseline, step_1, step_2, annual_kwh=(10000, 9000, 8000))
report = assess_renovation(step_2, settings, energy=energy)
show_table(report)
```

**Materials this step** counts only that renovation. **Materials so far** adds previous material impacts once. **Energy** describes the building after the step over 30 years. **Total** adds materials so far to that energy. **Savings** is baseline total minus this total; positive means a reduction.

Step 1: 220 + 9,000 × 30 × 0.1 = **27,220 kg CO₂e**; savings **2,780**. Step 2: 500 + 8,000 × 30 × 0.1 = **24,500**; savings **5,500**.

```python
import math

assert report.complete
assert math.isclose(report.stages[0].material_kgco2e, 220)
assert math.isclose(report.stages[1].material_kgco2e, 280)
assert math.isclose(report.stages[1].material_so_far_kgco2e, 500)
assert math.isclose(report.stages[0].savings_kgco2e, 2780)
print("Windows removed in step 1:", report.quantities(step=1, action="removed"))
assert report.quantities(step=2, action="removed") == []
```

```python
bundled = baseline.renovate(windows, insulation)
one_step = assess_renovation(
    bundled,
    teaching_settings(bundled),
    energy=supplied_energy(baseline, bundled, annual_kwh=(10000, 8000)),
)
assert math.isclose(one_step.stages[-1].total_kgco2e, report.stages[-1].total_kgco2e)
report.plot()
```

The final totals match because the material work and final energy are the same. Replacing windows again would add another installation and removal; it would not be treated as keeping the original windows.
