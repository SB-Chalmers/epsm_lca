# Windows and wall insulation in one step

Use the declared helper inputs in teaching_case.py. All climate factors and energy amounts below are invented to make the arithmetic easy to check. No energy simulation is run.

```python
from teaching_case import (
    make_baseline,
    renovation_options,
    show_table,
    supplied_energy,
    teaching_settings,
)

from epsm_lca import assess_renovation

baseline = make_baseline()
windows, insulation = renovation_options(baseline)
renovated = baseline.renovate(windows, insulation)
settings = teaching_settings(renovated)
energy = supplied_energy(baseline, renovated, annual_kwh=(10000, 8000))
report = assess_renovation(renovated, settings, energy=energy)
show_table(report)
```

Window production, transport and installation sum to 20 kg CO₂e/m². Removal, transport, treatment and disposal sum to 2 kg CO₂e/m². **10 × (20 + 2) = 220 kg CO₂e.**

New insulation weighs **40 m² × 0.1 m × 35 kg/m³ = 140 kg**. Its production, transport and installation sum to 2 kg CO₂e/kg: **140 × 2 = 280 kg CO₂e**. Together: **500 kg CO₂e**. Each factor and selected stage can be inspected in settings.

```python
import math

assert len(report.stages) == 1 and report.complete
assert math.isclose(report.stages[0].material_kgco2e, 500)
assert math.isclose(report.stages[0].total_kgco2e, 24500)
assert math.isclose(report.stages[0].savings_kgco2e, 5500)
assert not baseline.stages
print(settings.to_dict())
```

The supplied energy covers space heating and ventilation fans. Its climate factor is 0.1 kg CO₂e/kWh. Baseline: **10,000 × 30 × 0.1 = 30,000**. Renovated energy: **8,000 × 30 × 0.1 = 24,000**. Add 500 for renovation materials to get **24,500 kg CO₂e**, or **5,500 saved**.

```python
print("Removed quantities:", report.quantities(action="removed", by_material=True))
print("Added quantities:", report.quantities(action="added", by_material=True))
report.plot()
```
